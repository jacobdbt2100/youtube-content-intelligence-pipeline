# Creators — Silver Transformation

- Bronze → Silver
- Inspect Bronze creator data
- Transform to Silver schema
- Write Silver Delta table
- Validate Silver data

## 1. Inspect Bronze data

```python
%sql
SELECT *
FROM youtube_content_intelligence.bronze.brz_creators
LIMIT 5;
```

```python
%sql
DESCRIBE TABLE youtube_content_intelligence.bronze.brz_creators;
```

## 2. Write to Silver

```python
df_bronze = spark.table("youtube_content_intelligence.bronze.brz_creators")

df_bronze.write.format("delta").mode("overwrite").saveAsTable(
    "youtube_content_intelligence.silver.slv_creators"
)
```

## 3. Validate Silver table

```python
%sql
-- Preview Silver data
SELECT *
FROM youtube_content_intelligence.silver.slv_creators
LIMIT 5;
```

```python
%sql
-- Check row count
SELECT COUNT(*) AS row_count
FROM youtube_content_intelligence.silver.slv_creators;
```

```python
%sql
-- Check duplicate channel IDs
SELECT channel_id, COUNT(*) AS count
FROM youtube_content_intelligence.silver.slv_creators
GROUP BY channel_id
HAVING COUNT(*) > 1;
```
